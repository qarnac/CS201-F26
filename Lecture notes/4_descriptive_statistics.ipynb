# Descriptive Statistics

Descriptive statistics offers a way to summarize and describe relatively basic but essential features of a quantitative dataset. Watch this video for a quick overview of descriptive statistics: https://www.youtube.com/watch?v=SplCk-t1BeA

# Functions for Simple Descriptive Statistics

Python supports a few built-in functions for analyzing data in a list.

*   The `len` function returns the number of elements in a list.
*   The `sum` function returns the sum of all elements in a list. This will work well only when the elements in the list are numeric.
*   The `min` function returns the smallest value among the elements in a list.
*   The `max` function returns the smallest value among the elements in a list.

The following code block illustrates how to call the `len` function. Again, a **function call** starts with the name of the function, followed by a pair of parentheses. Within the parentheses, we need to identify the list variable we would like the function to analyze.

The following code block shows an example of calling the `len` function to determine the number of elements in the `days_of_a_week`.

```python
days_of_a_week = ['Sunday', 'Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday']
len(days_of_a_week)
```

We can also put the call to a function directly inside a print statement.

```python
print("Number of days:", len(days_of_a_week))
print(f'Number of days: {len(days_of_a_week)}')
```

As mentioned before Python compares strings of letters based on their **alphabetical** order. This is why, as shown by the following code block, when calling the `min` function with community_colleges, `'Grossmont'` is returned.

```python
community_colleges = ['MiraCosta', 'Grossmont', 'Miramar', 'Palomar']
min(community_colleges)
```

Therefore, it is more meaningful to determine the `min`, `max`, and `sum` for quantitative data. It is a common practice to save the result of a function using an assignment statement and then display the result using the print function. This is illustrated below.

```python
# data extracted from https://www.minimum-wage.org/wage-by-state
minimum_wages = [7.25, 11.73, 14.35, 11.00, 16.00, 14.42, 15.69, 13.25, 17.00, 7.25, 13.00, 7.25,
                 14.00, 7.25, 7.25, 7.25, 7.25, 7.25, 14.15, 15.00, 15.00, 10.33, 10.85, 7.25, 12.30,
                 10.30, 12.00, 12.00, 7.25, 15.13, 12.00, 15.00, 7.25, 7.25, 10.45, 7.25, 14.20, 7.25,
                 9.50, 14.00, 7.25, 11.20, 7.25, 7.25, 7.25, 13.67, 12.00, 16.28, 8.75, 7.25, 7.25]

number_of_states = len(minimum_wages)
lowest = min(minimum_wages)
highest = max(minimum_wages)
```

Running the above code block saves both the raw and aggregated data into respective variables. It does not generate any display. The following code block shows the values saved in the three variables respectively without any labels or explanations.

```python
number_of_states, lowest, highest
```

We can use the print function to add context for the values.

```python
print(f'The number of states is {number_of_states}')
print(f'The lowest minimum wage is ${lowest}')
print(f'The highest minimum wage is ${highest}')
```

We can combine list comprehension and the descriptive statistics functions to help us analyze a dataset.

```python
wages_under_12 = [item for item in minimum_wages if item < 12]
print("The number of states whose minimum wages are under $12 is:", len(wages_under_12))
```

# Measures of Central Tendency

**Measures of central tendency** describe the center or "middle section" of a list of numbers. Here are three common measures of central tendency:
*  **mean**: this is calculated by using the sum of all numbers divided by the count of all numbers.
*  **median**: this is the middlemost number in a sorted list.
*  **mode**: this is the most frequently occur number in a list.

The `statistics` module/library supports these functions. The following code block shows a template one would use to call these functions on a list.



```python
# Do not run this code block. It shows a template, not fully working code.
# Assuming stats has been designated as an alias for the statistics module.
# Assuming that we have already build list_variable to hold a list of numeric values.
# The mean or mathematical average for the list can be found through
stats.function_name(list_variable)
```

The following code block imports the statistics module/library and designate stats as its alias. It uses [unemployment data for San Marcos from YCHARTS](https://ycharts.com/indicators/san_marcos_ca_unemployment_rate#:~:text=San%20Marcos%2C%20CA%20Unemployment%20Rate%20is%20at%204.60%25%2C,lower%20than%20the%20long%20term%20average%20of%205.28%25.) to build a list.

```python
rates_2023 = [3.70, 3.60, 3.30, 3.00, 3.10, 3.80, 4.00, 4.30, 4.20, 4.30, 4.30, 4.30]
```

In the following code block, `stats.mean` tells Python that we are using the `mean` function inside the `statistics` library.

```python
import statistics as stats
stats.mean(rates_2023)
```

The median unemployment rate can be determined using the `median` function from the `statistics` library.

```python
stats.median(rates_2023)
```

The mode unemployment rate can be determined using the `mode` function from the `statistics` library.

```python
stats.mode(rates_2023)
```

We can save the above statistics into variables and use the print function to help display them in full sentences.

```python
import statistics as stats
rates_2023 = [3.70, 3.60, 3.30, 3.00, 3.10, 3.80, 4.00, 4.30, 4.20, 4.30, 4.30, 4.30]
rates_mean = stats.mean(rates_2023)
print(f'The average unemployment rates is {rates_mean: .2f}.')  #.2f indicates we want to show 2 digits after the decimal point

rates_median = stats.median(rates_2023)
print(f'The median unemployment rates is {rates_median: .2f}.')

rates_mode = stats.mode(rates_2023)
print(f'The most frequent monthly unemployment rates in 2023 is {rates_mode: .2f}.')
```

We can also use list comprehension to help us create a list of unemployment rates that are higher or lower than the meam or median.

```python
print("unemployment rates lower than the average:", [item for item in rates_2023 if item < rates_mean])
print("unemployment rates higher than the median:", [item for item in rates_2023 if item > rates_median])
```

# Measures of Dispersion

Measures of dispersion describe how tightly or loosely the data points cluster to the mean.
*  Low dispersion indicates the mean provides a relatively accurate representation of the "center" of the data.
*  High dispersion indicates we need to interpret the mean with caution.

Here are three populare measures of dispersion:
*  range: the difference between the largest and smallest number in the dataset.
*  stdev (standard deviation): how much each number in a dataset varies from the mean/average. As shown in the following curve, about 68% of the values in a normally distributed dataset are within one standard deviation from the mean.

![bell curve](https://openbooks.library.umass.edu/app/uploads/sites/26/2020/07/Bell-curve.jpg)

Let's go back to our unemployment rates list:

```python
rates_2023 = [3.70, 3.60, 3.30, 3.00, 3.10, 3.80, 4.00, 4.30, 4.20, 4.30, 4.30, 4.30]
```

Last time, we learned about the `min` and `max` function. We can use them to help us find the range of the unemployment rates as shown in the following code block. Note that these functions are not part of the statistics module.

```python
range = max(rates_2023) - min(rates_2023)
print(f'The range of unemployment rate is {range: .2f}')
```

The `statistics` module/library supports a `stddev` function to return the standard deviation among a list of data.

```python
import statistics as stats
stats.stdev(rates_2023)
```

We can save the above statistics into variables and then use print function to help explain the results.

```python
import statistics as stats
rates_2023 = [3.70, 3.60, 3.30, 3.00, 3.10, 3.80, 4.00, 4.30, 4.20, 4.30, 4.30, 4.30]

range = max(rates_2023) - min(rates_2023)
print(f'The difference between the highest and lowest unemployment rates is {range}.')

std = stats.stdev(rates_2023)
print(f'The standard deviation of the unemployment rates is {std: .3f}.')
```
