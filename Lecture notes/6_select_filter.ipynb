# Data Preparation

This code block creates a `DataFrame` object that will be used by other code blocks in this page. Please make sure that you run it before running other code blocks in this page.

```python
#source of data: https://www.pewresearch.org/social-trends/2014/04/10/next-america/#the-generational-divide
import pandas as pd
population_data = {'Year':[1960, 1970, 1980, 1990, 2000, 2010, 2020, 2030, 2040, 2050, 2060],
                  'White':[.85, .83, .80, .76, .69, .64, .60, .55, .51, .47, .43],
                  'Black': [.10, .11, .12, .12, .12, .12, .13, .13, .13, .13, .13],
                  'Hispanic': [.04, .05, .06, .09, .13, .16, .19, .22, .25, .28, .31],
                  'Asian': [.01, .01, .02, .03, .04, .05, .06, .06, .07, .08, .08],
                  'Others': [.00, .00, .01, .01, .02, .03, .03, .04, .04, .05, .06]}
population_df = pd.DataFrame(population_data)
population_df

#make sure that you run this code block before running the code blocks below
```

# Select column(s)

To access one column of a DataFrame object, we simply use the name of the DataFrame object followed by a pair of square brackets. Inside the square brackets, we add the name of the column. As a reminder, text is case sensitive in Python.

```python
population_df['White']
```

Each column of a DataFrame object is a Series object. DataFrame and Series share many attributes and functions. For example, the following code block shows how to access the info function for the 'White' column of population_df.

```python
population_df['White'].info()
```

```python
# This is to set the display format for floating point values
pd.options.display.float_format = "{:.2f}".format

# This is to save a column of a DataFrame into a variable
white = population_df['White']

# Then we can reference the column using the variable
print('the result of the shape attribute', white.shape)
print('the result of the describe function: is shown below')
white.describe()
```

# Filtering by One comparison

```python
population_df
```

The following code block creates a new DataFrame that contain rows whose White column have values that are under 0.5.

```python
white = population_df['White']
white_under_50_percent = population_df[white < 0.5]
white_under_50_percent
```

We can also specify the filtering condition beforehand as shown in the following code block.

```python
filtering_condition = population_df['White'] < 0.5
white_under_50_percent = population_df[filtering_condition]
white_under_50_percent
```

We can also combine the intermediate steps and achieve the same goal via the following.

```python
white_under_50_percent = population_df[population_df['White'] < 0.5]
white_under_50_percent
```

It is important to know that the filtering creates a new DataFrame without changing the original. Run the following code block to see the original DataFrame.

```python
population_df
```

The above filtering compares one column against a target value. We can also compare two columns as illustrated in the following code block.

```python
# Create a new DataFrame with rows whose Black percentage is higher than Hispanic percentage
black = population_df['Black']
hispanic = population_df['Hispanic']
filtering_condition = black > hispanic
# The above three lines can be rewritten into
# filtering_condition = population_df['Black'] > population_df['Hispanic']

more_black_than_hispanic = population_df[filtering_condition]
more_black_than_hispanic
```

# The & and | logical Operators

Similar to list comprehension, we can use the logical operators. However, pandas uses **& for and, | for or**. Since the &, and | operators have lower precedence than the comparison operators, it is important to put a pair of parentheses around the comparisons.

The following code block shows an example of filtering based on white values between 30% and 60%, i.e. 0.3 and 0.6.

```python
# Rows where the percentage for white is between 30 and 60 percent.
white = population_df['White']
filtering_condition = (white >= 0.3) & (white <= 0.6)
between_30_60_percent = population_df[filtering_condition]
between_30_60_percent
```

The following code block shows an example filtering values outside of the 30 - 60 range. That is, we are looking for white percentage either less than 30% or greater than 60%.

```python
# Rows where the percentage for white is either under 30 or above 60 percent.
filtering_condition = (population_df['White'] < 0.3) | (population_df['White'] > 0.6)
beyond_30_60_percent = population_df[filtering_condition]
beyond_30_60_percent
```

We can use the | operator to help us choose rows for selected years.

```python
filtering_condition = (population_df['Year'] == 1990) | (population_df['Year'] == 2010) | (population_df['Year'] == 2030)
years_1990_2010_2030 = population_df[filtering_condition]
years_1990_2010_2030
```

We can also use the & operator to help us exclude rows from selected years.

```python
filtering_condition = (population_df['Year'] != 1990) & (population_df['Year'] != 2010) & (population_df['Year'] != 2030)
years_not_1990_2010_2030 = population_df[filtering_condition]
years_not_1990_2010_2030
```

The comparisons connected by a logical operator can compare different columns.

```python
# Rows with high Black and Hispanic percentage
filtering_condition = (population_df['Black'] > 0.12) & (population_df['Hispanic'] > 0.12)
high_black_hispanic = population_df[filtering_condition]
high_black_hispanic
```

# The isin function and the ~ operator

In addition, pandas supports the **isin** function for that we can use to check if a given value belongs to a list of values.

The following code block shows an example of filtering based on the 'Year' column.

```python
selected_years = [1960, 1990, 2020, 2060]
condition = population_df['Year'].isin(selected_years)
from_selected_years = population_df[condition]
from_selected_years
```

The ~ operator is a logical operator that negate an original Boolean value. It can be used along with a condition to choose rows that do not meet the condition.

```python
filtering_condition = population_df['Year'].isin([1960, 1980, 2010, 2030, 2050])
outside_selected_years = population_df[~filtering_condition]
outside_selected_years
```
