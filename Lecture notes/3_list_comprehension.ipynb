# List comprehensions

List slicing allows us to create a new list by picking elements of an original list by their indices. List comprehensions offer a concise way to create new lists from existing ones based on an expression and an optional condition. They follow this general structure:

```python
# This is a template for list comprehension. Do not run the code block.
new_list = [expression for item in original_list if filtering_condition]
```

In the above structure:

*   `expression`: This is the operation performed on each `item` to determine the elements of the `new_list`. It can be the `item` itself or a transformation of the `item`. For now, we will focus on using the `item` itself.
*   `for`, `in`, and `if` are keywords in Python. Do not change them.
*   `new_list` should be a descriptive name for the sublist you are creating.
*   `item` is a variable that represents each element in the original list during iteration. You can choose a more descriptive name if needed.
*   `original_list` is the list you are iterating over.
*   `filtering_condition` is an optional condition that determines whether an `item` is included in the `new_list`. If omitted, all items from the `original_list` are processed.

Let's look at some examples using the following two lists, one for the ages of 10 students and the other for their respective class standings.

```python
ages = [18, 20, 27, 24, 29, 17, 23, 22, 21, 25]
standings = ['freshman', 'junior', 'senior', 'junior', 'senior', 'sophomore', 'senior', 'freshman', 'sophomore', 'freshman']
```

# Comparison operators

Python supports the following comparison operators that can be used to specify criteria in the `filtering_condition`:

*   Less than: `<`
*   Greater than: `>`
*   Less than or equal to: `<=`
*   Greater than or equal to: `>=`
*   Equal to: `==`
*   Not equal to: `!=`

In the following block, we are setting the filtering condition to be item < 20. That is, only items in the ages list that are under 20 will be used to create the new list.

```python
under_20 = [item for item in ages if item < 20]
under_20
```

If we want a new list to include ages that are above 25, we can use the following code block.

```python
above_25 = [item for item in ages if item > 25]
above_25
```

```python
print('original list is', ages)
print(f'under 20 {under_20}')
```

The comparison operators work for text strings too. In the following example, we create two new lists by comparing each item of the standings list against 'junior'.

```python
juniors = [item for item in standings if item == 'junior']
non_juniors = [item for item in standings if item != 'junior']
print("Juniors:", juniors)
print("Non-juniors:", non_juniors)
```

Keeping in mind, however, the comparisons will be based on alphabetical order and case sensitive.

Run the following code block to see what would be created for the two new lists.

```python
Seniors = [item for item in standings if item > 'junior']
Juniors = [item for item in standings if item == 'Junior']
print("Juniors", Juniors)
print("Seniors", Seniors)
```

The Juniors list is empty because the filtering condition was set to match "Junior" with an uppercase J. The Seniors list include sophormores because 'sophomore' is alphabetically larger than 'junior'.

# Logical Operators

In each of the above examples, we compare an item from the original list against one value. In addition to comparison operators, we can use logical operators (`and`, `or`, `not`) to combine multiple comparisons or negate a comparison. This allows for more complex filtering.

Let's see some examples using the `ages` and `standings` lists we defined earlier.

## The and operator

The **`and`** operator can connect two comparisons. It requires the item to satisfy both comparisons.

For example, the filtering condition in the following code block requires item to be at least 20 AND no more than 25.

```python
between_20_25 = [item for item in ages if item >= 20 and item <= 25]
between_20_25
```

In the following example, 'freshman' and 'junior' are in the new list because they satisfy both item != 'sophore' and item !='senior'

```python
freshmen_and_juniors = [item for item in standings if item != 'sophomore' and item != 'senior']
freshmen_and_juniors
```

## The or operator

The **`or`** operator also connects two comparisons. However, it allows the item to be included in the new list as long as it satisfies at least one of the two comparisons.

For example, in the following example, the **`or`** operator allows ages younger than 20 or older than 25 to be included in the new list.

```python
ages = [18, 20, 27, 24, 29, 17, 23, 22, 21, 25]
younger_or_older = [item for item in ages if item < 20 or item > 25]
younger_or_older
```

Similarly, in the following example,
*  'freshman' can be in the new list because it satisfies the comparison on the left of the **`or`** operator.
*  'senior' can be in the new list because it satisfies the comparison on the right of the **`or`** operator.

```python
freshmen_or_seniors = [item for item in standings if item == 'freshman' or item == 'senior']
print("Freshmen or seniors:", freshmen_or_seniors)
```
