## DataFrame and Series

Social scientists often use surveys to collect data for their research questions. Data collected from a survey are often stored in a **tabular** format where
*   Each **column** corresponds to a question on the survey and contains responses to the question from all respondants.
*   Each **row** corresponds to one respondant and covers responses for all questions from the respondant.

Similarly, a crime reporting data set is likely to be in a **tabular** format where
*   Each **column** corresponds to one aspect of a crime and contains details for that aspect from all crimes reported.
*   Each **row** corresponds to one crime and covers details for all aspects of the crime.

`DataFrame` and `Series` are two popular modules from the `pandas` library. They share a lot of functions.
*  We will be using a `DataFrame` objects to hold one **tabular** dataset.
*  One **row** or one **column** of a `DataFrame` would be considered as a `Series` object.

# Create a DataFrame Object

The following code block illustrates how to create a `DataFrame` object. Run the code to see how the rows and columns of the `DataFrame` object are organized.

The `population_datat` variable uses a **dictionary** object to set up the data and their structure for a `DataFrame` object. A dictionary uses a pair of curly brackets to enclose a list of **key:value** pairs.
*   The **key** before each colon specifies the name of a column.
*   The **value** after each colon specifies the **list** of values to be stored under the related column heading. The number of elements in the list dictate the number of rows for the table. When creating a dataset for a `DataFrame` object, it is important that **all lists share the same number of elements**.

Line 3 demonstrates how a new `DataFrame` object can be created using the `pd.DataFrame` function. It takes a dictionary object (`population_dataset`) as its argument. The assignment statement store the new `DataFrame` object into the variable `population_df`.

```python
#source of data: https://www.pewresearch.org/social-trends/2014/04/10/next-america/#the-generational-divide
import pandas as pd
population_data = {
    'Year':[1960, 1970, 1980, 1990, 2000, 2010, 2020, 2030, 2040, 2050, 2060],
    'White':[.85, .83, .80, .76, .69, .64, .60, .55, .51, .47, .43],
    'Black': [.10, .11, .12, .12, .12, .12, .13, .13, .13, .13, .13],
    'Hispanic': [.04, .05, .06, .09, .13, .16, .19, .22, .25, .28, .31],
    'Asian': [.01, .01, .02, .03, .04, .05, .06, .06, .07, .08, .08],
    'Others': [.00, .00, .01, .01, .02, .03, .03, .04, .04, .05, .06]
    }
population_df = pd.DataFrame(population_data)
population_df

#make sure that you run this code block before running the code blocks below
```

We can also use the print function to show the content of a DataFrame object.

```python
print(population_df)
```

Here is another example for creating a new DataFrame object.
*  'category', 'planned', and 'actual' are the names for the columns.
*  A colon is used to separate a column name and the list of values for that column.
*  The items of the lists match by their positions in the list. They make up the respective rows of the table. For example, the first row of the table has 'Rent' for the 'category' column, 950 for the 'planned' column, and 950 for the 'actual' column.

# Initial inspections

Before performing any data analysis or cleaning, it's crucial to perform an initial inspection of your pandas DataFrame. This helps you understand its structure, content, data types, and identify potential issues such as missing values or incorrect data formats.

In the following subsections, we will introduce and demonstrate the use of several useful pandas functions for this initial inspection.

## The head and tail function

The `head()` function is used to display the first few rows of a DataFrame. By default, it shows the first 5 rows. You can specify the number of rows to display by passing an integer argument to the function. This is useful for getting a quick look at the structure and content of your data.

```python
population_df.head()
```

```python
population_df.head(3)
```

We can save the result of the head() function into a new DataFrame object.

```python
first_3_years = population_df.head(3)
print(first_3_years)
```

## The tail() function

The `tail()` function is similar to `head()`, but it displays the *last* few rows of a DataFrame. By default, it also shows the last 5 rows. You can specify the number of rows to display by passing an integer argument to the function. This is useful for checking the end of your dataset and can sometimes reveal patterns or issues at the end of the data collection.

```python
population_df.tail()
```

Demonstrate how to use the `tail` function to display a specific number of rows of the `population_df` DataFrame.


```python
population_df.tail(3)
```

We can save the result of the tail function into a new DataFrame object.

```python
last_3_years = population_df.tail(3)
print(last_3_years)
```

## The info() function

The `info()` function provides a concise summary of a DataFrame. This includes:
*   The index dtype and columns.
*   Non-null values and their counts for each column.
*   Data types (dtypes) for each column.
*   Memory usage of the DataFrame.

This function is essential for a quick overview of your data's structure, helping you identify missing values (columns with fewer non-null entries than the total number of rows) and check if the data types are as expected.

```python
population_df.info()
```

Here are a few things we can learn about the dataset from the above result of the info function.

*  "RangeIndex: 11 entries, 0 to 10": This means there are 11 rows in the dataset. The index for the first row is 0. The index for the last row is 10.
*  "Data columns (total 6 columns)": This means there are 6 columns in the data set.
*  " 0   Year      11 non-null     int64": This means that the first column's name is 'Year'. There are 11 non-null values in this column. The data type for the values in the column is int64, i.e. integer.
*  "1   White     11 non-null     float64": This means that the second column's name is 'White'. There are 11 non-null values in this column. The data type for the values in the column is float64, i.e. floating point values.
*  "memory usage: 660.0 bytes": This indicates that the dataset takes up 660 bytes of memory.

```python
budget_df.info()
```

# The describe() function

The `describe()` function generates descriptive statistics of the DataFrame's numerical columns. This includes count, mean, standard deviation, minimum, maximum, and quartile values. This function is very useful for getting a statistical summary of your data and identifying potential outliers or distributions.

```python
population_df.describe()
```

There are too many digits after the decimal points, making the results hard to understand. We can use pd.options.display.float_format = "{:.2f}".format to help specify the number of digits after the decimal place.

```python
pd.options.display.float_format = "{:.2f}".format
population_df.describe()
```

Here are some observations we can make based on the above descriptive statistics:

*  The **count** row shows that there are 11 rows in the table.
*  The **min** row shows the smallest value for each column. The earliest data is for 1960. The lowest percentage for white is 0.43, i.e. 43% while the lowest percentage for Hispanic is 0.04, i.e. 4%. It is important to note that we do not know from the above descriptive statistics which year these percentages belong to.
*  The **50%** row shows the median value for each column. For example, 50% of the time, White accounts for at most 64% of the population.
*  The **max** row shows the largest value for each column. The latest data is for 2060, as a prediction. The highest percentage for Black is 13%, while the highest percentage for Asian is 8%. Again, we do not know which year these percentags belong to by simply looking at these statistics.
*  The **mean** row shows the average value for each column. On average, White has higher percentage than other ethnicities.
*  The **std** row shows the standard deviation for each column. The small std value for Black indicates not much change in their percentage. The higher std value for White indicates more change in their percentage.
