# Ranking

The `describe` function show the smallest and highest values of incarceration and poverty. However, it does not tell us which states have these values. **Ranking** assigns an order position to each row based on the values of a numeric column. This can be achieved using the `rank` function as shown in the following code block.

*  The `rank` function does not change any value in `df['Incarceration']`.
*  It does not rearrange the rows of the DataFrame object.
*  It simply creates a new `Series` that can be assigned to a new column of the DataFrame object.
*  `ascending=False` indicates that the highest value will get first rank. If we want the lowest value to get the first rank, we should set `ascending=True`.
*  `method='dense'` indicates that there will be no gaps in ranks and ties share a rank. If we want tied values to get the best (lowest) rank and the next rank skips, we will need to use `method='min'`

```python
df['Incarceration_rank'] = df['Incarceration'].rank(ascending=False, method='dense')
df
```

If we want the lowest value to receive 1 as its rank, we need to set `ascending=True` as shown in the following code block.

```python
df['Incarceration_rank'] = df['Incarceration'].rank(ascending=True, method='dense')
df
```

```python
condition = df['Incarceration_rank'] < 5
df[condition]
```

# Sorting

Adding the rank columns do not change the order of the original rows. Sorting, on the other hand, creates a new DataFrame object where the rows are rearranged rows based on column values.

```python
df.sort_values(by='Incarceration', ascending=True)
```

By default, the sort_values function creates a new DataFrame object without changing the original DataFrame object. Run the following code block and you'll see that the original DataFrame object remained unsorted.

```python
df
```

There are two ways to rearrange the rows of the orginal DataFrame object.

The first, and preferred way, is to simply assigned the newly created DataFrame object by the sort_values function back to the original DataFrame object.

```python
df = df.sort_values(by='Incarceration', ascending=False)
df
```

Another way, discouraged by the pandas community, is to use inplace=True

```python
df.sort_values(by='Incarceration', ascending=True, inplace=True)
df
```

# Counting

Last time, we learned about setting up filtering conditions to extract specific rows. When applying a condition to a column, pandas will evaluate each value in the column and returns True if the value meets the condition, False if the value does not meet the condition.

```python
df
```

```python
filtering_condition = (df['Incarceration'] > 5.5) & (df['Incarceration'] <= 7.1)
filtering_condition
```

As shown by the above result, there are three rows that return True for the condition. In Python, and many programming languages, True is consider 1 and False is consider 0. Therefore, we can use the sum function to help us add up the values of 1 as a way to count the number of values in a column that meet the condition.

```python
filtering_condition.sum()
```

The np in the above result is a common alias for NumPy, another popular third party Python library. The result np.int64(3) indicates that there are 3 rows in the dataset that meets the filtering condition.

The following code block shows more examples of using filtering conditions to help count related rows.

```python
under_condition = df['Incarceration'] <= 5.5
above_condition = df['Incarceration'] > 7.1
between_condition = (df['Incarceration'] > 5.5) & (df['Incarceration'] <= 7.1)
outside_condition = (df['Incarceration'] <= 5.5) | (df['Incarceration'] > 7.1)
under_count = under_condition.sum()
above_count = above_condition.sum()
between_count = between_condition.sum()
outside_count = outside_condition.sum()
print(f'There are {under_count} states with incarceration under 5.5')
print(f'There are {above_count} states with incarceration above 7.1')
print(f'There are {between_count} states with incarceration between 5.5 and 7.1')
print(f'There are {outside_count} states with incarceration outside of the range of 5.5 and 7.1')
```

# Classify numeric data by ranges

The maps on HDPulse use different shades of blue to help people visualize the incarceration and poverty levels. We are not quite ready to learn how to generate such maps. However, it is actually quite easy to classify numeric values into categories based on ranges.

The `cut` function takes a few parameters:

*  The first parameter is the numeric column for which we wan to classify.
*  The `bins` parameter specifies the ranges that will be used to break down the numeric values. By default, the ranges are lower/left exclusive and upper/right inclusive.  
*  The `labels` parameter specifies the labels for respective ranges.

The following two code blocks show different ways of classifying incarceration rates.

```python
# This will mark values > 1.6 to 4.1 as low, >4.1 to 5.5 as medium low,
# > 5.5 to 7.1 as medium, > 7.1 to 8.6 as medium high, and >8.6 to 12.6 as high
# Right/upper inclusive means 4.1 belongs to the first bin
# Left/lower exclusive means 4.1 does not belong to the second bin
incarceration_bins = [1.6, 4.1, 5.5, 7.1, 8.6, 12.6]
incarceration_labels = ['low', 'medium low', 'medium', 'medium high', 'high']
df['Incarceration_Level'] = pd.cut(df['Incarceration'],
                                   bins = incarceration_bins,
                                   labels = incarceration_labels)
df
```

We can make the bins left/lower inclusive and right/upper exclusive by setting the right parameter to False as shown in the following code block.

*  Left/lower inclusive means 1.6 belong to the first bin and 5.6 belongs to the second bin.
*  Right/upper exclusive means 5.6 does not belong to the first bin.

```python
# This will mark values 1.6 to <5.6 as low, 5.6 to <8.7 as medium, 8.7 to <12.6 as high
incarceration_bins = [1.6, 5.5, 8.6, 12.6]
incarceration_labels = ['low', 'medium', 'high']
df['Incarceration_Level'] = pd.cut(df['Incarceration'],
                                   bins = incarceration_bins,
                                   labels = incarceration_labels,
                                   right=False)
df
```
