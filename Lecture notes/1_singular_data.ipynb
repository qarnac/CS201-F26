# Variables



The term **variable** refers to the name we use to identify a piece of data our program will be working with. Each name may contain letters, underscores, or digits but cannot start with a digit. It is important to give each variable a descriptive name to make it easy for us to remember what the data is about. Here are example valid variable names. When more than one words are used for a variable name, it is common to use an underscore to connect them. Do not use space or other special characters in a variable name.
*  name
*  age
*  salary
*  first_name
*  last_name
*  age_limit
*  annual_salary

There are some words that Python has already used them for specific purpose. They are referred to as **reserved words** or **keywords**. Do not use them as your variable names.

# The print function

In Python, a **function** is defined to carry out a specific task. There are many functions available and we will help you learn how to use them as they become handy for your tasks. When we use a function in our program, we say we make a **function call** or we call the function.

To call a function, we start with the name of the function, followed by a pair of parentheses. Within the parentheses, we provide information needed for the function to carry out its tasks. We call the items inside the parentheses **parameters** or **arguments**.

We will be using many functions to perform data related tasks.



The `print` function is designed to help display information. When calling the `print` function, the pair of parenthese after the function name is used to enclose the items to be displayed. By default, each `print` function displays one line. The following code block shows three example calls to the `print` function.

*   A `print` function call with nothing inside the parentheses will display an empty line.
*   A `print` function call with a specific value inside the parentheses will display the value.
*   A `print` function call with a variable inside the parentheses will display the value assigned to the variable. If you didn't run the code block above, Python will not know what to display for `county`.

```python
print()
print('California')
print()
print(county)
```

Each `print` function call can include multiple items inside the parentheses by separating them using commas. Text within quotes will be displayed as-is but Python will display the values of variables provided.

```python
print("We're interested in the", county, 'county.')
print('The current average gas price for', gas_rating, 'octane is', current_average)
```

***Printting via f-string***

Python offers a powerful feature called **f-strings** (formatted string literals) to provide a concise and intuitive way to embed variables directly into strings. To create an **f-string**, we simply prefix the string with the letter f. Within the string, we can then use a pair of **curly bracket** to include a variable.

For example, in the following code block, we use easily use f-string to embed two variables into one sentence.

```python
state = 'California'
data = 331.4
print(f'According to the 2020 Census, the population for {state} reached {data} million.')
```
