# Data Preparation

As a reminder, I've created an updated dataset that contains selected demographics columns and the average values for respective sections.
Refer to the [VCQ code book](https://docs.google.com/document/d/1uYfUuUsPVQyD3puVvbIMjamYuVPr_f4gtJuSB5Wj-WA/edit) for questions in related sections.

Run the following code block to create the DataFrame object and a correlation heatmaps on selected sections.

```python
import pandas as pd
pd.options.display.float_format = '{:.2f}'.format

df = pd.read_csv('https://raw.githubusercontent.com/qarnac/cs201/refs/heads/main/vcq_recoded_means.csv')

cols = ['stress', 'esteem', 'family_satisfaction', 'friend_satisfaction', 'life_satisfaction']
corr_matrix = df[cols].corr()

import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(5, 4))

sns.heatmap(data=corr_matrix,
    annot=True,      # show correlation values
    cmap='coolwarm' # blue-red color scale
)

plt.title('Correlation Heatmap on Selected Sections')
plt.show()
```
