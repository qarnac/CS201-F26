# Data preparation

In this assignment, you will analyze data from the World Happiness Report, an international survey that measures how people evaluate the quality of their lives. Respondents are asked to rate their lives on a scale from 0 to 10, where 0 represents the worst possible life and 10 represents the best possible life. The average rating for each country is reported as a happiness score.

Run the following code block to collect data from a [simplified CSV file](https://raw.githubusercontent.com/qarnac/cs201/refs/heads/main/world_happiness_report.csv) I created.

```python
import pandas as pd
url = "https://raw.githubusercontent.com/qarnac/cs201/refs/heads/main/world_happiness_report.csv"
df = pd.read_csv(url)
df.info()
```

The dataset has 9 columns and 149 rows. Each row describes information about one country:
*  `Country name`: Name of a country
*  `Regional indicator`: Which region the country belong
*  `Happiness`: Average happiness score (0 to 10)
*  `GDP`: Logged GDP per capita - a measure of economic prosperity (7 to 12), larger value indicate wealthier countries.
*  `Social_support`: Measure of social support available to individuals (0 to 1), , percentage of respondents answering yes to having relatives or friends they can count on.
*  `Life_expectancy`: Healthy life expectancy (50 to 80)
*  `Freedom`: Freedom to make life choices (0 to 1), percentage of respondents giving a positive response for recognizing freedom to make choices.
*  `Generosity`: Charitable donations adjusted for GDP (-0.3 to 0.7)
*  `Corruption`: Perceptions of corruption (0 to 1), higher values inicate a larger proportion of people perceiving corruption.
