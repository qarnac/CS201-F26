# Data preparation

I have downloaded [Homicide & Legal Intervention Death Rates for 2023](https://tinyurl.com/hdpulsehomicide2023) data by states and created a CSV file at
https://raw.githubusercontent.com/qarnac/cs201/refs/heads/main/homicide_reduced_states.csv. Not all states have data broken down by gender. Therefore, not all states are included in the CSV file.

Here are the columns of the CSV file:

*  State: this column holds the names of the states whose homicide & legal intervention death rates are available for both female and male.
*  All_Rate: this column holds the deaths per 100,000 for corresponding states, regardless of sex.
*  All_Count: this column holds the number of deaths for corresponding states, regardless of sex.
*  Female_Rate: this column holds the female deaths per 100,000 for corresponding states.
*  Female_Count: this column holds the number of female deaths for corresponding states.
*  Male_Rate: this column holds the female deaths per 100,000 for corresponding states.
*  Male_Count: this column holds the number of female deaths for corresponding states.
