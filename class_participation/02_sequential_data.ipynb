# Indexing Elements in a List

Behind the scene, Python numbers the elements in a list through an **indexing** system that is based on how much the element **offsets** from the first element. That is

*   The **index** for the first element of a list is 0 since there is no offset.
*   The **index** for the second element of a list is 1 since it is one off from the first element.
*   The **index** for the third element of a list is 2 since it is two off from the first element.
*   The **index** for the 10th element of a list is 9 since it is nine off from the first element.

Accessing a single element of a list can be done through the name of the list followed by a pair of [ ]. Inside the [ ], include the index for the element you want to access.

For example, in the following code block, me[0] refers to 'Professor', me[1] refers to 'Computer Science', me[2] refers to 'Science II', and me[3] is 219.

```python
me = ['Professor', 'Computer Science', 'Science II', 219]
print("The first element of me is", me[0])
print("The second element of me is", me[1])
print("The third element of me is", me[2])
print("The fourth element of me is", me[3])
```

# List Slicing

List slicing allows us to access a range of elements from a list. It is commonly done through this format:
![list slicing](https://www.learnbyexample.org/wp-content/uploads/python/Python-List-Slicing-Syntax.png)


*  L is the name of a list
*  `start`: The index at which the slice begins (inclusive). If omitted, the slice starts from the beginning of the list.
*  `end`: The index at which the slice ends (exclusive). If omitted, the slice goes up to the end of the list.
*  `step`: The interval between elements in the slice. If omitted, the default step is 1.

Given the following list of colors:


```python
colors = ['red', 'orange', 'yellow', 'green', 'blue', 'indigo', 'violet']
colors
#It is important to run this code block before running other code blocks for list slicing.
```

`colors[0]` is `'red'`, `colors[1]` is `'orange'`, and so on. `'violet'` is `colors[6]`.

If we want to slice out `['red', 'orange', 'yellow']`, we can use `colors[0 : 3]` or `colors[ : 3]`.

*  Start slicing at index 0.
*  The value of `end` is 3, meaning that the slicing will not include `colors[3]`.

Run the following code block to verify.

```python
colors[ : 3]    # Starts from the beginning and goes up to but not include index 3
```

If we want to slice out `['blue', 'indigo', 'violet']` from `colors`, we can use `colors[4 : ] `since the index for `'blue'` is 4 and `'violet'` is the last in `colors`. When the `end_index` is omitted, it means going all the way till the end of the list. Run the following code block to verify. Change 4 to other numbers to see how that affect the slice.

```python
colors[4 : ]  #Starts from index 4 and goes all the way to the end of the list
```

If we want to slice out `['yellow', 'green', 'blue']` from `colors`, we can use `colors[2 : 5]` since the index for `'yellow'` is 2 and the index for `'blue'` is 4. As a reminder `colors[2 : 5]` means to include `colors[2]`, `colors[3]`, and `colors[4]` but not `colors[5]`.

Run the following code block to verify. Change the numbers to see how that affect the slice.

```python
colors[2 : 5] #Starts from index 2 and goes up to but not include index 5
```

We only add `: step` to the brackets if we want to skip elements. For example, if we want to slice out `['orange', 'green', 'indigo']`, we can use `colors[1 : 6 : 2]`. It indicates that slicing will start at `colors[1]`, adding 2 to the index to get the next element, i.e. `colors[3]`, until it reaches `colors[6]`. Again `colors[6]` will not be included. Run the following code block to verify. Use a different step number to see how that affect the slice.

```python
colors[1 : 6 : 2]  #Starts at index 1, adds 2 each time until it goes beyond 6
```

What do you think the following slicing will extract? Think about it and then run the code block to verify.
